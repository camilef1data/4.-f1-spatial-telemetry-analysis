```python
import matplotlib.pyplot as plt
import numpy as np

# Métricas de comparação
metricas = ['Vitórias', 'Pódios', 'Pole Positions', 'Voltas Rápidas']

# Dados preenchidos (sem erro de sintaxe)
dados_leclerc = [4, 11, 5, 3]
dados_hamilton = [3, 12, 2, 4]

x = np.arange(len(metricas))
largura = 0.35

fig, ax = plt.subplots(figsize=(10, 6))

# Barras lado a lado com as cores da Ferrari
barras_lec = ax.bar(x - largura/2, dados_leclerc, largura, label='Charles Leclerc', color='#A60010')
barras_ham = ax.bar(x + largura/2, dados_hamilton, largura, label='Lewis Hamilton', color='#1a1a1a')

ax.set_title('Scuderia Ferrari: Comparativo de Performance Interna', fontsize=14, fontweight='bold', pad=15)
ax.set_ylabel('Quantidade Acumulada')
ax.set_xticks(x)
ax.set_xticklabels(metricas)
ax.grid(axis='y', linestyle=':', alpha=0.6)
ax.legend()

# Coloca os números em cima de cada barra
for barras in [barras_lec, barras_ham]:
    for barra in barras:
        altura = barra.get_height()
        ax.annotate(f'{altura}',
                    xy=(barra.get_x() + barra.get_width() / 2, altura),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Posições de chegada nas corridas
posicoes = ['1º (Vitória)', '2º Lugar', '3º Lugar', '4º ao 5º', '6º ao 10º', 'Fora da Zona/DNF']
frequencia = [3, 5, 4, 5, 2, 1]

plt.figure(figsize=(9, 5))

# Tons de laranja da McLaren
cores_mclaren = ['#FF8000', '#ff9933', '#ffb366', '#ffcc99', '#ffe6cc', '#737373']
plt.barh(posicoes, frequencia, color=cores_mclaren, edgecolor='black', alpha=0.9)

# Coloca o 1º lugar no topo
plt.gca().invert_yaxis()

plt.title("McLaren Racing: Análise de Consistência de Lando Norris", fontsize=13, fontweight='bold', pad=15)
plt.xlabel("Número de Corridas")
plt.grid(axis='x', linestyle='--', alpha=0.5)

# Adiciona os números na ponta das barras
for index, value in enumerate(frequencia):
    plt.text(value + 0.1, index, str(value), va='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
