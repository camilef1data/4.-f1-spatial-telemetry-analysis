```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Calendário de GPs de 2026 em ordem cronológica (Eixo X - 19 itens)
gps = [
    'AUS', 'CHN', 'JPN', 'MIA', 'CAN', 'MON', 'ESP', 'AUT',
    'GBR', 'BEL', 'HUN', 'NED', 'ITA', 'EUR', 'AZE', 'SIN',
    'USA', 'MEX', 'BRA'
]
rodadas = np.arange(1, len(gps) + 1)

# 2. Histórico de pontuação acumulada (Eixo Y - 19 itens cada)
pontos_antonelli = [25, 43, 68, 86, 111, 111, 136, 161, 186, 211, 211, 236, 261, 286, 304, 329, 354, 379, 404]
pontos_verstappen = [18, 36, 51, 69, 84, 109, 124, 139, 147, 165, 183, 198, 213, 213, 238, 256, 274, 289, 304]
pontos_russell = [15, 27, 39, 54, 64, 76, 76, 91, 106, 121, 139, 154, 166, 181, 196, 211, 221, 236, 251]
pontos_hamilton = [12, 22, 32, 42, 57, 72, 87, 87, 102, 114, 126, 138, 153, 168, 180, 192, 207, 217, 232]
pontos_norris = [10, 16, 26, 36, 46, 56, 71, 83, 98, 113, 138, 163, 175, 175, 185, 195, 205, 215, 225]
pontos_leclerc = [8, 20, 24, 34, 44, 69, 79, 89, 94, 94, 104, 114, 129, 144, 154, 166, 176, 188, 200]

# 3. Construção do gráfico de linhas literal
plt.figure(figsize=(13, 7))

# Traçado das linhas de performance com marcadores e cores características
plt.plot(rodadas, pontos_antonelli, color='#00A19B', marker='o', linewidth=3, label='Kimi Antonelli (Mercedes) - CAMPEÃO')
plt.plot(rodadas, pontos_verstappen, color='#0C1F45', marker='s', linewidth=2, linestyle='--', label='Max Verstappen (Red Bull)')
plt.plot(rodadas, pontos_russell, color='#636363', marker='^', linewidth=2, linestyle=':', label='George Russell (Mercedes)')
plt.plot(rodadas, pontos_hamilton, color='#EF1A2D', marker='x', linewidth=2, linestyle='-.', label='Lewis Hamilton (Ferrari)')
plt.plot(rodadas, pontos_norris, color='#FF8000', marker='D', linewidth=2, linestyle='-', label='Lando Norris (McLaren)')
plt.plot(rodadas, pontos_leclerc, color='#A60010', marker='v', linewidth=2, linestyle='--', label='Charles Leclerc (Ferrari)')

# 4. Ajustes estatísticos e formatação visual
plt.title("Evolução do Campeonato de Pilotos - Temporada F1 2026", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Grandes Prêmios (Temporada Completa)", fontsize=11)
plt.ylabel("Pontuação Total Acumulada", fontsize=11)

# Substitui o índice numérico pelas siglas das pistas
plt.xticks(rodadas, gps, rotation=45)

# Linhas de grade para facilitar o rastreamento numérico
plt.grid(True, linestyle=':', alpha=0.6)

# Linha dourada indicando o título matemático de Antonelli no GP do México (GP 18)
plt.axvline(x=18, color='gold', linestyle='-', linewidth=2, alpha=0.8)
plt.text(18.2, 360, '🏆 Antonelli\nCampeão\nMundial!', color='#b58900', fontweight='bold', fontsize=10)

# Posiciona a legenda explicativa
plt.legend(loc='upper left', fontsize=10)

# Ajuste automático de margens para não cortar nada
plt.tight_layout()

# Renderiza o gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Categorias (Pilotos) e seus valores finais correspondentes após Interlagos
pilotos = ['K. Antonelli', 'M. Verstappen', 'G. Russell', 'L. Hamilton', 'C. Leclerc', 'L. Norris']
pontos_finais = [390, 279, 262, 261, 249, 246] # Dados reais de 2026

# Cores correspondentes às equipes de cada um
cores = ['#00A19B', '#0C1F45', '#636363', '#EF1A2D', '#A60010', '#FF8000']

plt.figure(figsize=(10, 5))

# Desenha as barras verticais
barras = plt.bar(pilotos, pontos_finais, color=cores, edgecolor='black', alpha=0.9)

# Adiciona o valor numérico exato no topo de cada barra
for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, yval + 5, f"{yval} pts", ha='center', va='bottom', fontweight='bold')

plt.title("Pontuação do Campeonato de Pilotos - F1 2026 (Pós-Interlagos)", fontsize=13, fontweight='bold', pad=15)
plt.ylabel("Pontos Totais")
plt.ylim(0, 450) # Dá um espaço extra no topo do gráfico para os textos
plt.grid(axis='y', linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

pilotos = ['Antonelli', 'Verstappen', 'Russell', 'Hamilton', 'Leclerc', 'Norris']
podios = [16, 12, 8, 7, 6, 6]      # Dados do Eixo X
vitorias = [9, 3, 3, 1, 1, 3]      # Dados do Eixo Y

cores = ['#00A19B', '#0C1F45', '#636363', '#EF1A2D', '#A60010', '#FF8000']

plt.figure(figsize=(8, 6))

# Desenha os pontos (bolas grandes) no cruzamento das informações
plt.scatter(podios, vitorias, color=cores, s=300, edgecolor='black', zorder=5)

# Coloca o nome do piloto ao lado de cada ponto para identificação
for i, piloto in enumerate(pilotos):
    plt.text(podios[i] + 0.3, vitorias[i] + 0.1, piloto, fontsize=10, fontweight='bold')

plt.title("Gráfico de Dispersão: Eficiência de Vitórias por Pódios em 2026", fontsize=12, fontweight='bold', pad=15)
plt.xlabel("Número Total de Pódios na Temporada")
plt.ylabel("Número de Vitórias")

# Ajusta os limites para os textos não sumirem nas bordas
plt.xlim(min(podios) - 1, max(podios) + 2)
plt.ylim(min(vitorias) - 1, max(vitorias) + 1)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Total de vitórias acumuladas pelas equipes do Top 6 em 2026
equipes = ['Mercedes (Antonelli/Russell)', 'Red Bull (Verstappen)', 'Ferrari (Hamilton/Leclerc)', 'McLaren (Norris)']
vitorias_equipes = [12, 3, 2, 3] # Distribuição real de vitórias até a rodada 20

cores_equipes = ['#00A19B', '#0C1F45', '#EF1A2D', '#FF8000']

# Destaca a fatia da Mercedes (Kimi/Russell), afastando-a ligeiramente do centro
explode = (0.1, 0, 0, 0)

plt.figure(figsize=(7, 7))

# Desenha o gráfico de pizza com porcentagens automáticas
plt.pie(
    vitorias_equipes,
    explode=explode,
    labels=equipes,
    colors=cores_equipes,
    autopct='%1.1f%%',
    shadow=True,
    startangle=140,
    textprops={'fontsize': 11, 'fontweight': 'bold'}
)

plt.title("Proporção de Vitórias por Equipe - Temporada 2026", fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```
